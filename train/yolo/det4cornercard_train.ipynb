```python
from ultralytics import YOLO
import torch
```

```python

# Load a model
model = YOLO("/home/tuanndq/ocr_cccd/yolo/model/detect/yolo11n.pt")  # load a pretrained model (recommended for training)

# Train the model với cấu hình GPU
results = model.train(
    data="data/v2/dataset_yolov11_det4cornercard/dataset.yaml",
    epochs=300,
    imgsz=640,
    device=0,  # Sử dụng GPU đầu tiên (0), hoặc 'cpu' nếu muốn dùng CPU
    batch=32,  # Điều chỉnh batch size phù hợp với GPU memory
    workers=4,  # Số worker threads
    verbose=True,
    amp=False,
)
```

```python
# Load a model
model = YOLO("runs/detect/train_det4cornercard/weights/best.onnx")  # load a custom model
```

```python
# Validate the model
metrics = model.val()  # no arguments needed, dataset and settings remembered
metrics.box.map  # map50-95
metrics.box.map50  # map50
metrics.box.map75  # map75
metrics.box.maps  # a list contains map50-95 of each category
```

```python

# Run batched inference on a list of images
results = model.predict(
    [
        "data/v2/dataset_yolov11_det4cornercard/val/images/905_1_VTM_20250307153512_125.jpg",
        "data/v2/dataset_yolov11_det4cornercard/val/images/UPDATE_LICENSE_20250220_220956.704_fLa_ofl.jpg",
        "data/v2/dataset_yolov11_det4cornercard/val/images/EKYC_BACK_1723087215171.jpg",
        "data/v2/dataset_yolov11_det4cornercard/val/images/EKYC_Android_20240723_050400.000_hsd_ofl.jpg",
        "data/v2/dataset_yolov11_det4cornercard/val/images/EKYC_Android_20240824_032054.054_SzP_obl.jpg",
        "data/v2/dataset_yolov11_det4cornercard/val/images/EKYC_Android_20250609_022916.016_jWn_obl.jpg",
        "data/v2/dataset_yolov11_det4cornercard/val/images/EKYC_BACK_1723120446669.jpg",
        "data/v2/dataset_yolov11_det4cornercard/val/images/DEP_308_1_VTM_20250320194601_380.jpg",
        "data/imgtest/EKYC_Android_20250517_031847.047_UQX_ofl.jpg"
    ],
    conf = 0.3,
    iou = 0.1,
    imgsz=640,

    # stream=True
)  # return a generator of Results objects

# Process results generator
for i,result in enumerate(results):
    boxes = result.boxes  # Boxes object for bounding box outputs
    masks = result.masks  # Masks object for segmentation masks outputs
    keypoints = result.keypoints  # Keypoints object for pose outputs
    probs = result.probs  # Probs object for classification outputs
    obb = result.obb  # Oriented boxes object for OBB outputs
    # result.show()  # display to screen
    result.save(filename=f"test_inference/det4cornercard/det4cornercard_{i}.jpg")  # save to disk
```
